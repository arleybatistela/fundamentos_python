# **Exercício 1 - Calculadora de IMC**

```python
# 1. Entrada de dados do usuário (convertendo para número decimal float)
peso = float(input("Digite o seu peso em kg (ex: 70.5): "))
altura = float(input("Digite a sua altura em metros (ex: 1.75): "))

# 2. Cálculo do IMC (peso dividido pela altura ao quadrado)
imc = peso / (altura * altura)

# 3. Exibição do valor do IMC formatado com 2 casas decimais
print(f"\nSeu IMC é: {imc:.2f}")

# 4. Verificação da faixa de peso usando estruturas condicionais (if/elif/else)
if imc < 18.5:
    print("Classificação: Abaixo do peso")
elif imc < 25.0:
    print("Classificação: Peso normal")
elif imc < 30.0:
    print("Classificação: Acima do peso")
elif imc < 35.0:
    print("Classificação: Obesidade")
elif imc < 40.0:
    print("Classificação: Obesidade nível 2")
else:
    print("Classificação: Obesidade nível 3")
```

# **Exercício 2 - Cumprimentar pelo Primeiro Nome**

```python
# 1. Pergunta o nome completo do usuário
nome_completo = input("Digite seu nome completo: ")

# 2. O método .split() divide o nome em uma lista de palavras
# O [0] pega a primeira palavra da lista (o primeiro nome)
primeiro_nome = nome_completo.strip().split()[0]

# 3. Exibe o cumprimento utilizando o primeiro nome
print(f"Olá, {primeiro_nome}! Seja muito bem-vindo(a)!")
```

# **Exercício 3 - Extrair o Domínio do E-mail**

```python
# 1. Solicita que o usuário digite o e-mail
email = input("Digite o seu e-mail: ")

# 2. O .split('@') divide o e-mail em duas partes: antes e depois do '@'
# O [1] pega a segunda parte (que vem depois do '@', ou seja, o domínio)
dominio = email.strip().split('@')[1]

# 3. Imprime o resultado final conforme o exemplo solicitado
print(f"O domínio do seu e-mail é {dominio}")

```

# **Exercício 5 - Folha de Pagamento e Descontos**

```python
# 1. Entrada de dados
ganho_por_hora = float(input("Quanto você ganha por hora? R$ "))
horas_trabalhadas = float(input("Quantas horas você trabalhou no mês? "))

# 2. Cálculo do Salário Bruto
salario_bruto = ganho_por_hora * horas_trabalhadas

# 3. Cálculo dos descontos
imposto_renda = salario_bruto * 0.11  # 11% de IR
inss = salario_bruto * 0.08           # 8% de INSS
sindicato = salario_bruto * 0.05      # 5% para o Sindicato

# 4. Cálculo do total de descontos e do Salário Líquido
total_descontos = imposto_renda + inss + sindicato
salario_liquido = salario_bruto - total_descontos

# 5. Exibição dos resultados
print("\n--- Relatório do Salário ---")
print(f"+ Salário Bruto    : R$ {salario_bruto:.2f}")
print(f"- IR (11%)         : R$ {imposto_renda:.2f}")
print(f"- INSS (8%)        : R$ {inss:.2f}")
print(f"- Sindicato (5%)   : R$ {sindicato:.2f}")
print(f"= Salário Líquido  : R$ {salario_liquido:.2f}")
```

# **Exercício 6 - Ordenar 3 Números em Ordem Crescente**

```python
# 1. Entrada de 3 números
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
num3 = float(input("Digite o terceiro número: "))

# 2. Criamos uma lista com os 3 números
numeros = [num1, num2, num3]

# 3. O método .sort() organiza a lista em ordem crescente
numeros.sort()

# 4. Convertendo para inteiros se não tiverem casas decimais (apenas para exibição mais limpa)
n1 = int(numeros[0]) if numeros[0].is_integer() else numeros[0]
n2 = int(numeros[1]) if numeros[1].is_integer() else numeros[1]
n3 = int(numeros[2]) if numeros[2].is_integer() else numeros[2]

# 5. Exibição da resposta formatada conforme o exemplo
print(f"\nOs números digitados em ordem crescente foram {n1}, {n2} e {n3}.")
```

# **Exercício 7 - Verificar se o Ano é Bissexto**

```python
# 1. Solicita ao usuário que digite um ano
ano = int(input("Digite um ano para verificar: "))

# 2. Regra do Ano Bissexto:
# Um ano é bissexto se for divisível por 4 E não for divisível por 100,
# OU se for divisível por 400.
# O operador % (módulo) pega o resto da divisão. Se o resto for 0, é divisível.

if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    print(f"O ano {ano} é BISSEXTO!")
else:
    print(f"O ano {ano} NÃO é bissexto.")
```
