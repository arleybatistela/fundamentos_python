# **Fatiamentos de String**

```python
minha_string = "hello world"

#Acessar caracteres individualmente

print(minha_string[0]) # h
print(minha_string[1]) # e
print(minha_string[2])# l
print(minha_string[3])# l
print(minha_string[4]) # o

#Concatenar strings

outra_string = " World"
nova_string = minha_string + outra_string
print(nova_string)

# Repetir strings

repetida = minha_string*20
print(repetida)

# Fatiar strings
print(minha_string[1:5])
print(minha_string[:4])
print(minha_string[:5])
print(minha_string[:])
print(minha_string[:2])
print(minha_string[2:])
print(minha_string[  :  : 2]) # Aqui ele pega de 2 em 2 casas
print(minha_string[-5])
print(minha_string[-1])
print(minha_string[::-1])

print(type(minha_string))
```
