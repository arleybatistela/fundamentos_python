### **Laços de Repetição**

```python
count = 0
while count <= 5:
    print(f"O valor do contador é {count}")
    count += 1

    #print("Aqui o while acabou !!")
```

```python
count = 0
while count < 10:
    print(f"O valor do contador é {count}")
    count += 1
    #continue --- Aqui com o continue o loop volta no while e apresente o n, depois faz o loop e volta.

    
    print("teste do continue")
    count +=200

print("Aqui o while acabou !!")
```

```python
import time

count = 0

while True:
    if count <= 10:
        print(f"O valor do contador é {count}")
        count += 1
    else:
        print("Aqui executa antes do break ")    
        break

print("Carregando...")    
time.sleep(3)

print(" Tô forado while, o break me quebrou gente kkkk ")
    
```

### **Laço de repetição FOR**

```python
for i in range(-11, 11+1, 2):
    print(i)
```

```python
nomes = ["Joao", "Maria","Ana"]
for nome in nomes:
    print(nome)
```

```python
nome = "João"
for letra in nome:
    print(letra)
```
