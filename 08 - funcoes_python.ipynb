### **Função em Python**

### Sintaxe da Função:

def nome_função (Aqui eu posso ou não ter argumentos/parâmentros):
    '''
        Aqui dentro eu coloco a documentação da minha função  ###  **É uma explicação para o usuário quando pegar essa função p/ executar**
    '''    

    # A sua função realiza as coisas aqui

    return # Pode ou nao ter o return que serve para retornar algo onde chamamos a função


```python
# Função sem argumento e sem return

def say_hello():
    print("Hello People") # Aqui criei a função.

say_hello() # Aqui chamei a função.

say_hello()
say_hello()
say_hello()
```

```python
# função com argumento

def greeting(name):
    print(f"Hello {name}")

nome = input("Digite seu nome")

greeting(nome)   
```

```python
# função imprima o maior número

def maior (x,y):
    if x > y:
        print(x)
    else:
        print(y)

maior(10, 15)    
```

```python
# função usando return

def add_num(num1, num2):
    return num1+num2

add_num(10,5)
```
